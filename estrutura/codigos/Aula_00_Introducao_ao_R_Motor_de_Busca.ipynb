## Módulo 1 — Rodar, atribuir, e o corpus como vetor
- `<-` atribui valores;
- Quase tudo em R é vetor (um número ou string isolada é um vetor de tamanho 1);
- **R começa em índice 1** (índice 0 devolve vetor vazio);
- Operações são vetorizadas por padrão;
- `length()` conta elementos do vetor.


```r
docs <- c(                                                          # c() = combine: cria vetor; <- atribui
  d1 = "recuperacao de informacao ordena documentos por relevancia", # nome = texto
  d2 = "o modelo de espaco vetorial representa documentos como vetores",
  d3 = "bm25 e um modelo probabilistico de ranqueamento de texto",
  d4 = "aprendizado estatistico fundamenta a recuperacao moderna",
  d5 = "o indice invertido acelera a busca em muitos documentos",
  d6 = "embeddings capturam a semantica de palavras e documentos",
  d7 = "a avaliacao mede a relevancia dos resultados da busca",
  d8 = "ciencia de dados combina estatistica e programacao"
)
length(docs)
```

## Módulo 4 — Texto II: `strsplit` e `unlist`
- `strsplit` aceita múltiplos textos, por isso sempre retorna uma **lista**;
- `unlist` achata a lista devolvendo um vetor simples.


```r
frase <- substr(docs[["d1"]], 1, 25)
strsplit(frase, " ")         # devolve lista com [[1]]
unlist(strsplit(frase, " ")) # achata para vetor simples
```

## Módulo 5 — Funções: `tokenizar`
- Funções são criadas com `function(...)`;
- Não exigem `return`: a última expressão avaliada é automaticamente devolvida;
- Variáveis criadas dentro da função têm escopo estritamente local.


```r
tokenizar <- function(texto) {
  texto <- tolower(texto)
  unlist(strsplit(texto, " "))
}
tokenizar(frase)
tokenizar("Ciencia de DADOS")
```

**Checkpoint 5:**
Função `maiuscula` que extrai a primeira palavra em maiúsculas.


```r
maiuscula <- function(texto) toupper(unlist(strsplit(texto, " "))[1])
maiuscula(docs[["d3"]])
```

## Módulo 6 — `lapply` e `sapply`
- `lapply`: aplica uma função a cada item de uma lista/vetor e **sempre devolve uma lista** (essencial quando cada saída tem tamanho diferente);
- `sapply`: aplica e **tenta simplificar** para um vetor ou matriz quando os resultados têm dimensões homogêneas.


```r
tokens <- lapply(docs, tokenizar)
length(tokens)
tokens[["d4"]]
sapply(tokens, length) # simplifica para vetor nomeado
sum(sapply(tokens, length)) # total de palavras no corpus
```

## Módulo 7 — `table`, `factor`, `sort`, `%in%`
- `table`: calcula a contagem de frequência de cada termo;
- `factor(..., levels = vocab)`: força as contagens a respeitarem uma ordem fixa pré-definida e inclui zeros para os termos ausentes (permitindo que todos os documentos gerem vetores de tamanho idêntico);
- `%in%`: testa pertencimento; negação com `!` permite filtrar stopwords.


```r
table(tokens[["d3"]])["de"]
sort(table(tokens[["d3"]]), decreasing = TRUE)[1:3]

vocab <- c("de", "modelo", "busca")
table(factor(tokens[["d3"]], levels = vocab)) # note a inclusao do termo ausente 'busca' com contagem 0!
```

```r
# Filtragem de termos:
tokens[["d1"]][!tokens[["d1"]] %in% c("de", "por")]
```

## Módulo 8 — Matrizes e reciclagem
- Em R, matrizes são preenchidas **por coluna** por padrão;
- A multiplicação de uma matriz por um vetor (`m * peso`) aplica a mecânica de **reciclagem**: o vetor é repetido ao longo de cada coluna, permitindo ponderar cada termo/linha sem qualquer laço.


```r
m <- matrix(c(1, 0, 1, 1), nrow = 2)
rownames(m) <- c("de", "modelo")
colnames(m) <- c("d1", "d2")
m

peso <- c(1, 10)
m * peso # linha 'de' multiplicada por 1, linha 'modelo' por 10
```

```r
rowSums(m)
colSums(m)
```

## Módulo 9 — Regex I — procurar: `grep`, `grepl`, `^`, `$`, `|`, `[ ]`, `+`
- `grep` retorna índices/posições dos elementos compatíveis;
- `grepl` retorna vetor booleano (`TRUE`/`FALSE`);
- Âncoras: `^` (início), `$` (fim);
- **Regex casa pedaços, não palavras completas**: `grep("de", docs)` casa "moderna" e "mede".


```r
grep("modelo", docs)
grepl("modelo", docs)
names(docs)[grepl("modelo", docs)]

grep("^o ", docs)          # inicia com "o "
grep("busca$", docs)       # termina em "busca"
grep("busca|modelo", docs) # alternancia: busca OU modelo
grep("[0-9]", docs)        # contem digito
```

```r
# Diferenca crucial:
grep("de", docs)   # acha d4 e d7 porque estao dentro de "moderna" e "mede"
grep(" de ", docs) # acha apenas a palavra isolada entre espacos
```

**Checkpoint 9:**
- `grep("documentos$", docs)` -> `5 6`
- `grep("^e", docs)` -> `6` (d6 começa com "embeddings")


```r
grep("documentos$", docs)
grep("^e", docs)
```

## Módulo 10 — Regex II — transformar e limpar: `sub`, `gsub`, `.`, `*`, `{n}`, `[^ ]`, `trimws`
- `sub`: altera a primeira ocorrência;
- `gsub`: altera globalmente todas as ocorrências;
- `trimws`: remove espaços das extremidades;
- Pipeline de higienização de texto em 4 passos;
- `tokenizar` definitiva com `\s+`.


```r
sub("de", "DE", docs[["d3"]])    # troca a primeira (em moDElo!)
gsub("de", "DE", docs[["d3"]])   # troca todas
gsub("[aeiou]", "", docs[["d1"]]) # apaga vogais
gsub("[0-9]+", "", docs[["d3"]])  # apaga digitos
```

```r
# O Pipeline de Higienizacao:
sujo <- "  O Modelo   de Espaco   Vetorial!! "
x <- trimws(sujo)
x <- tolower(x)
x <- gsub("[^a-z ]", "", x)
x <- gsub("\s+", " ", x)
x
```

```r
# Funcao de tokenizacao robusta:
tokenizar <- function(texto) unlist(strsplit(tolower(texto), "\\s+"))
tokenizar(sujo)
```

### Módulo 12 — Três frases próprias: sujar, limpar, contar
Criação de um vetor com 3 frases intencionalmente ruidosas sobre a Baixada Santista e o Porto de Santos, aplicação do pipeline de 4 linhas e análise de frequência.


```r
f <- c(
  f1 = "  O Porto   de Santos bateu recorde de movimentacao de cargas em 2026!! ",
  f2 = "A operacao portuaria de Santos movimenta cargas, conteineres e navios... ",
  f3 = "Santos planeja novos investimentos na infraestrutura de logistica portuaria! "
)
f
```

```r
# Pipeline de limpeza em 4 linhas:
x_f <- trimws(f)
x_f <- tolower(x_f)
x_f <- gsub("[^a-z ]", "", x_f)
x_f <- gsub("\\s+", " ", x_f)
x_f
```

```r
tokens_f <- lapply(x_f, tokenizar)
freq_f   <- table(unlist(tokens_f))
sort(freq_f, decreasing = TRUE)

cat("Total de tokens:", length(unlist(tokens_f)), "\n")
cat("Tokens unicos (vocabulario):", length(unique(unlist(tokens_f))), "\n")
```
